```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from typing import Tuple

def load_embeddings(root_directory: str) -> Tuple[np.ndarray, np.ndarray]:
    """
    Load image embeddings and class labels from .npy files.

    Args:
        root_directory (str): The directory containing the .npy files.

    Returns:
        Tuple[np.ndarray, np.ndarray]: Image embeddings and class labels.
    """
    embeddings_path = f"{root_directory}/image_embeddings.npy"
    labels_path = f"{root_directory}/class_labels.npy"
    
    image_embeddings = np.load(embeddings_path)
    class_labels = np.load(labels_path)
    
    return image_embeddings, class_labels

def visualize_embeddings(image_embeddings: np.ndarray, class_labels: np.ndarray) -> None:
    """
    Visualize image embeddings using PCA and a scatter plot.

    Args:
        image_embeddings (np.ndarray): The image embeddings to visualize.
        class_labels (np.ndarray): The corresponding class labels.
    """
    # Reduce dimensionality to 2D using PCA
    pca = PCA(n_components=2)
    embeddings_2d = pca.fit_transform(image_embeddings)

    # Create a scatter plot
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(embeddings_2d[:, 0], embeddings_2d[:, 1], c=class_labels, cmap='viridis')
    plt.colorbar(scatter)
    plt.title('Image Embeddings Visualization (PCA)')
    plt.xlabel('First Principal Component')
    plt.ylabel('Second Principal Component')
    plt.show()

# Usage example
if __name__ == "__main__":
    root_directory = "..\\data\\normalized_3d_test"
    image_embeddings, class_labels = load_embeddings(root_directory)
    visualize_embeddings(image_embeddings, class_labels)
```
