```python
import plotly.graph_objects as go
import numpy as np

# Generate sample data
x = np.random.rand(10)
y = np.random.rand(10)
z = np.random.rand(10)

# Create the figure
fig = go.Figure()
```

```python
for i in range(len(x)):
    fig.add_trace(go.Scatter3d(
        x=[x[i]], y=[y[i]], z=[z[i]],
        mode='markers',
        marker=dict(
            symbol='square',
            size=40,
            color='rgba(0,0,0,0)',
            line=dict(width=0)
        ),
        text=f'Image {i}',
        hoverinfo='text'
    ))
    
    fig.add_layout_image(
        dict(
            source="https://www.google.com/images/branding/googlelogo/1x/googlelogo_color_272x92dp.png",
            xref="x", yref="y",
            x=x[i], y=y[i],
            sizex=0.1, sizey=0.1,
            sizing="contain",
            opacity=1,
            layer="above"
        )
    )
```

```python
fig.update_layout(
    scene=dict(
        xaxis=dict(range=[0, 1]),
        yaxis=dict(range=[0, 1]),
        zaxis=dict(range=[0, 1]),
    ),
    width=800,
    height=800,
    margin=dict(r=0, l=0, b=0, t=0)
)
```
